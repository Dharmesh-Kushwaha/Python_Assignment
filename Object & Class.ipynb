```python
class Student:
    def addDeatils(self,name,age,course):# Local Variables/Scope Variables
        self.n=name
        self.a=age
        self.c=course
    def showDetails(self):
        print("Name: ",self.n)
        print("Age: ",self.a)
        print("Course: ",self.c)

obj1=Student()
obj1.addDeatils("Amit",23,"Python")
obj1.showDetails()

obj2=Student()
obj2.addDeatils("Priya",20, "Java")
obj2.showDetails()
```
